## Interactive Equivalent Width Measurements for Stellar Spectra

SpelacEW is an interactive Python tool designed for measuring Equivalent Widths (EW) in stellar spectra using automatic Gaussian fitting.

The code supports:

- Interactive continuum selection
- Automatic Gaussian fitting
- Multi-Gaussian blending analysis
- Manual exploration mode
- Automatic PDF and CSV export
- Solar reference comparison

This tutorial presents the basic workflow and usage of the code.

````python
from spelacew import EW
````

## Required Input Files

SpelacEW requires:

1. A stellar spectrum
2. A line list in CSV format

Optionally, a solar reference EW file can also be provided.

---

### Supported spectrum formats

- `.fits`
- `.txt`
- `.dat`
- `.csv`

The spectrum must contain two columns:

- Wavelength
- Flux

---

### Line list format

The line list must contain at least the wavelength column.

Example:

| wavelength | element | species | ep | gf |
|---|---|---|---|---|
| 6151.617 | Fe I | 26 | 2.18 | -3.299 |

---

### Optional solar reference file

The solar reference CSV allows direct comparison between measured EWs and solar EWs.

## Running SpelacEW

The program can be executed by creating an `EW` object with:

- spectrum file
- line list
- optional solar reference
- plotting width

Example:

````python
EW(
    "spectrum.fits",
    "lines.csv",
    "solar.csv",
    width=2.0
)
````

## Main Interface

After executing the code, the interactive window will appear.

The interface contains:

- Main spectral panel
- Automatic zoom panel
- Central wavelength marker
- Optional solar reference continuum
- Interactive crosshair
- Fit diagnostics panel

The current line wavelength and element are displayed in the title.

## Navigating Between Spectral Lines

The user can navigate through the line list interactively.

| Key | Action |
|---|---|
| `n` | Next line |
| `p` | Previous line |

The selected spectral line is always centered in the main panel.

![Normal Fit](tutorial_images/main.png)

# Measuring an Equivalent Width

To measure a spectral line:

1. Press `k` on the left continuum point
2. Press `k` on the right continuum point

The code automatically:

- Builds the local continuum
- Normalizes the spectrum
- Fits a Gaussian absorption profile
- Computes the Equivalent Width (EW)
- Computes the FWHM
- Estimates the reduced χ²

![Normal Fit](tutorial_images/basic_fit.png)

## Fit Diagnostics

The fit results displayed on screen include:

- Central wavelength (`μ`)
- Equivalent Width (`EW`)
- Full Width at Half Maximum (`FWHM`)
- Reduced chi-square (`χ²`)

The right panel displays a zoomed view of the fitted region.

### Step 1 — Activate blending mode

Press:

`b`

### Step 2 — Define fitting region

Press `d` twice to define the fitting interval.

### Step 3 — Add Gaussian centers

Press `g` over each visible spectral component.

![Normal Fit](tutorial_images/blendingmode_1.png)

### Step 4 — Run the fit

Press:

`Enter`

![Normal Fit](tutorial_images/blendingmode_2.png)

# Output Files

SpelacEW automatically generates:

- A CSV table with measured parameters
- A PDF document containing all fit figures

## CSV Output

The `resultados.csv` file contains:

| Column | Description |
|---|---|
| wavelength | Central wavelength |
| wave_left | Left continuum point |
| wave_right | Right continuum point |
| left_continuum | Left continuum flux |
| right_continuum | Right continuum flux |
| ew | Measured Equivalent Width |
| ew_ref | Solar reference EW |
| FWHM | Gaussian FWHM |
| Chi2R | Reduced chi-square |
| element | Atomic species |
| ep | Excitation potential |
| gf | Oscillator strength |

## PDF Output

The generated PDF stores all successful fits during the session.

This allows fast visual inspection and posterior quality control of the measurements.

# Practical Recommendations

- Select continuum points away from the line core
- Avoid noisy spectral regions
- Use blending mode for asymmetric or merged lines
- Inspect χ² values to identify problematic fits
- Adjust the width parameter when nearby lines contaminate the fit
- Verify the automatic Gaussian center in blended regions
